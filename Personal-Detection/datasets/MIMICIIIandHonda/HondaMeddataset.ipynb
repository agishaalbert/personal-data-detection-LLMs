```python
import pandas as pd
import numpy as np

def print_unique_values(df):
    # Identify non-numeric columns (categorical and other non-numeric types)
    categorical_columns = df.select_dtypes(exclude=['int64', 'float64']).columns

    print("Unique values in each categorical column:")
    print("=========================================")

    for col in categorical_columns:
        unique_values = df[col].unique()
        nan_count = df[col].isna().sum()
        
        print(f"\nColumn: {col}")
        print(f"Number of unique values: {len(unique_values)}")
        print(f"Number of NaN values: {nan_count}")
        print("Unique values:")
        
        # Print each unique value
        for value in unique_values:
            if pd.isna(value):
                print("- NaN")
            else:
                print(f"- {value}")

# Usage
# Assuming you have a DataFrame named 'df'
print_unique_values(MIMICIIIdata)
```

```python
import pandas as pd
import numpy as np

def sample_with_unique_values(df, n=200):
    # Identify categorical columns (exclude numeric and datetime types)
    categorical_columns = df.select_dtypes(include=['object', 'category']).columns
    
    # Initial random sample
    sample = df.sample(n=n, random_state=42)
    
    # Check and add missing unique values
    for col in categorical_columns:
        unique_values = df[col].unique()
        sample_values = sample[col].unique()
        missing_values = set(unique_values) - set(sample_values)
        
        for value in missing_values:
            if pd.isna(value):
                row_to_add = df[df[col].isna()].sample(n=1)
            else:
                row_to_add = df[df[col] == value].sample(n=1)
            sample = pd.concat([sample, row_to_add])
    
    # If we've added rows, we need to randomly select again to get back to n
    if len(sample) > n:
        sample = sample.sample(n=n, random_state=42)
    
    return sample

# Function to check if all unique values are represented
def check_unique_values(original_df, sample_df):
    categorical_columns = original_df.select_dtypes(include=['object', 'category']).columns
    
    print("Checking unique values representation:")
    print("======================================")
    
    for col in categorical_columns:
        original_unique = set(original_df[col].unique())
        sample_unique = set(sample_df[col].unique())
        missing = original_unique - sample_unique
        
        print(f"\nColumn: {col}")
        print(f"Total unique values: {len(original_unique)}")
        print(f"Represented in sample: {len(sample_unique)}")
        print(f"Missing values: {len(missing)}")
        if missing:
            print("Missing unique values:")
            for value in missing:
                print(f"- {value}")

# Usage
result = sample_with_unique_values(MIMICIIIdata, n=200)
check_unique_values(MIMICIIIdata, result)
```

```python
def label_columns(df):
    personal_columns = {
        # Direct identifiers
        "subject_id", "hadm_id", "icustay_id", "dob", "dod", "dod_hosp", "dod_ssn", "gender",
        "insurance", "marital_status", "ethnicity",
        
        # Time-related identifiers
        "intime", "outtime", "chartdate", "charttime", "transfertime", "storetime", "starttime", "endtime",
        "createtime", "updatetime", "admittime", "dischtime", "deathtime", "edregtime", "edouttime",
        
        # Health and location-related information
        "diagnosis", "hospital_expire_flag", "patientweight", "location", "locationcategory", "first_careunit", "last_careunit",
        
        # Identifiers & Codes
        "orderid", "linkorderid", "icd9_code", "cpt_cd", "cpt_number", "ticket_id_seq", "drg_code",
        
        # Medication & Treatment Info
        "drug", "drug_name_generic", "formulary_drug_cd", "dose_val_rx", "route"
    }
    column_labels = {}
    for column in df.columns:
        if column in personal_columns:
            column_labels[column] = "personal"
        else:
            column_labels[column] = "non-personal"

    df_labels = pd.DataFrame(list(column_labels.items()), columns=['Features', 'label'], index=None)
    return df_labels['label']
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer, IterativeImputer
from sklearn.experimental import enable_iterative_imputer

# Assuming MIMICIIIdata is already loaded

# Separate numeric and non-numeric columns
numeric_columns = MIMICIIIdata.select_dtypes(include=[np.number]).columns
categorical_columns = MIMICIIIdata.select_dtypes(exclude=[np.number]).columns

# Handle columns with all missing values
all_missing = MIMICIIIdata[numeric_columns].columns[MIMICIIIdata[numeric_columns].isnull().all()].tolist()
numeric_columns = numeric_columns.drop(all_missing)

# Impute numeric columns
numeric_imputer = IterativeImputer(random_state=0, max_iter=10)
imputed_numeric = numeric_imputer.fit_transform(MIMICIIIdata[numeric_columns])
MIMICIIIdata[numeric_columns] = imputed_numeric

# Fill all-missing numeric columns with a placeholder value
for col in all_missing:
    MIMICIIIdata[col].fillna(-999, inplace=True)

# Impute categorical columns
categorical_imputer = SimpleImputer(strategy='most_frequent')
MIMICIIIdata[categorical_columns] = categorical_imputer.fit_transform(MIMICIIIdata[categorical_columns])

# Check for any remaining missing values
missing_values = MIMICIIIdata.isnull().sum()
print("Columns with missing values after imputation:")
print(missing_values[missing_values > 0])

# If there are still missing values, fill them with a placeholder
MIMICIIIdata.fillna('Unknown', inplace=True)

print("Imputation complete.")

MIMICIIIdata.to_csv('MIMICdata_imputed.csv')
```
