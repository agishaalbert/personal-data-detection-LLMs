```python
# Define the mapping of classes to "personal" and "non-personal"
class_mapping = {
    "Other data": "non-personal",
    "Phone_number": "personal",
    "Address": "personal",
    "Person": "personal",
    "Email": "personal",
    "NIN": "personal",
    "Date": "non-personal",
    "Organization": "non-personal",
    "GPE": "non-personal",
    "Geolocation": "non-personal",
    "SWIFT/BIC": "personal",
    "IBAN": "personal",
    "Passport": "personal",
    "Religion": "personal",
    "CCN": "personal",
    "ID_Card": "personal",
    "Sexuality": "personal",
    "Gender": "personal",
    "Nationality": "personal",
    "Race": "personal"
}

# # Function to classify a given class or combination of classes into "personal" or "non-personal"
# def classify_data(data_classes):
#     if not isinstance(data_classes, str):  # Ensure the input is a string
#         return "non-personal"
#     classes = data_classes.split(",")  # Split in case of multiple labels
#     for cls in classes:
#         if class_mapping.get(cls.strip(), "non-personal") == "personal":
#             return "personal"
#     return "non-personal"

def classify_data(data_classes):
    if data_classes is None:
        return "non-personal"  # Explicitly handle missing values
    data_classes = str(data_classes)  # Convert to string if possible
    classes = data_classes.split(",")  
    for cls in classes:
        if class_mapping.get(cls.strip(), "non-personal") == "personal":
            return "personal"
    return "non-personal"


# Function to classify an existing DataFrame
def classify_dataframe(df):
    df["Label"] = df["label"].apply(classify_data)
    df.rename(columns={"label": "Original labels"}, inplace=True)
    return df
```
